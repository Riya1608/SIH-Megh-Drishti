```python
import os

print("🔎 Searching for CARTO / map configuration...\n")

for root, dirs, files in os.walk("/content/Megh_drishti-main"):
    for file in files:
        if file.endswith((".js", ".html", ".jsx", ".tsx", ".py", ".css")):
            path = os.path.join(root, file)
            try:
                with open(path, "r", encoding="utf-8", errors="ignore") as f:
                    text = f.read()

                if "carto" in text.lower() or "basemap" in text.lower() or "tileLayer" in text:
                    print("📄", path)
                    for i, line in enumerate(text.splitlines(), 1):
                        if any(x in line.lower() for x in ["carto", "basemap", "tilelayer"]):
                            print(f"   Line {i}: {line.strip()}")
            except:
                pass
```
